# Portfolio Weight Calculation and Visualization

```python
#assigning weights

market_cap_data = {ticker: yf.Ticker(ticker).info['marketCap'] for ticker in tickers}

total_market_cap = sum(market_cap_data.values())

weights = {ticker: market_cap / total_market_cap for ticker, market_cap in market_cap_data.items()}
```

```python
weights
```

```python
weighted_returns_portfolio = returns.mul(weights,axis=1)
weighted_returns_portfolio.dropna()
px.line(weighted_returns_portfolio)
```

```python
# Display weights
weights_df = pd.DataFrame(list(weights.items()), columns=['Ticker', 'Weight'])
fig = px.bar(weights_df, x='Ticker', y='Weight', title='Portfolio Weights by Market Capitalization')
fig.show()
```

# Portfolio Returns Calculation and Visualization

```python
returns['Portfolio'] =weighted_returns_portfolio.sum(axis=1).dropna()
returns['Portfolio'] = returns['Portfolio'] * 100
fig=px.line(returns)
fig.show()
```

```python
# Calculate and visualize portfolio returns
returns['Portfolio'] = weighted_returns_portfolio.sum(axis=1)
returns['Portfolio'] = returns['Portfolio'] * 100

fig = px.line(returns, y='Portfolio', title='Portfolio Returns')
fig.show()
```

#Value at Risk (VaR) Calculation

```python
var95 = np.percentile(returns['Portfolio'], 5)
var99 = np.percentile(returns['Portfolio'], 1)

# VaR for the next 10 days
var99_10days = var99 * np.sqrt(10)

# Display VaR values
print(f'VaR at 95% confidence interval: {var95}')
print(f'VaR at 99% confidence interval: {var99}')
print(f'VaR at 99% confidence interval for the next 10 days: {var99_10days}')
```

```python
from scipy.stats import norm

weights_array = np.array(weights)
cov_matrix = returns.cov() # covariance matrix of returns
cov_matrix.head()
```

```python
avg_returns_array = avg_returns.values[:-1]  # Remove the last element from avg_returns
weights_array = np.array(list(weights.values()))
cov_matrix_trimmed = cov_matrix.iloc[:-1, :-1]  # Remove the last row and column from cov_matrix

portfolio_mean = np.dot(avg_returns_array, weights_array)
portfolio_std = np.sqrt(np.dot(weights_array.T, np.dot(cov_matrix_trimmed, weights_array)))
```

```python
confidence_level= 0.05 #calculate var for 95% confidence levl
VaR = norm.ppf(confidence_level, portfolio_mean, portfolio_std)
VaR
```
